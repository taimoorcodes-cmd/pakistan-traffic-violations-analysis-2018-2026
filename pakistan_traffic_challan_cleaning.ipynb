```python
# Convert Date from object to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Verify
print(df['Date'].dtype)
```

```python
# Check unique values
print("Cities:", df['City'].nunique())
print("Provinces:", df['Province'].nunique())
print("Violation Types:", df['Violation_Type'].nunique())
print("Vehicle Types:", df['Vehicle_Type'].nunique())
print("Payment Status:", df['Payment_Status'].unique())
print("Enforcement Method:", df['Enforcement_Method'].unique())
```

```python
# Check fine amount statistics
print(df['Fine_Amount_PKR'].describe())
```

```python
# Check duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check if Challan_ID is unique
print("Total rows:", len(df))
print("Unique Challan IDs:", df['Challan_ID'].nunique())
```

```python
# Check collected amount statistics
print(df['Amount_Collected_PKR'].describe())

# Check unpaid challans
unpaid = df[df['Payment_Status'] == 'Unpaid']['Amount_Collected_PKR'].sum()
print("Total unpaid amount:", unpaid)
```

```python
# Create fine category
df['Fine_Category'] = pd.cut(
    df['Fine_Amount_PKR'],
    bins=[0, 1000, 2000, 3000, 5500],
    labels=['Low', 'Medium', 'High', 'Very High']
)

print(df['Fine_Category'].value_counts())
```

```python
# Create payment binary column
df['Is_Paid'] = df['Payment_Status'].apply(
    lambda x: 1 if x == 'Paid' else 0
)

print(df['Is_Paid'].value_counts())
```

```python
# Final check
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
```
